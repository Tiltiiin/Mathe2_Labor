```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(figsize=(5, 5))
x = [-3,0,3,np.nan,2,-2]
y = [-3,6,-3,np.nan,0,0]
ax.plot(x,y,color ='black')
x_y = np.array([x, y])
alpha = np.radians(20)
rotate_array = np.array([[np.cos(alpha),-1*np.sin(alpha)],[np.sin(alpha),np.cos(alpha)]])
rotated = np.array(rotate_array.dot(x_y))
ax.plot(rotated[0], rotated[1], color='red')
x_max = 4.5
x_min = -4
y_max = 7.5
y_min = -5
ax.set_xticks(np.arange(x_min, x_max + 1, 1))
ax.set_yticks(np.arange(y_min, y_max + 1, 1))
ax.set_xticks(np.arange(x_min + 0.5, x_max, 1.0), minor=True)
ax.set_yticks(np.arange(y_min + 0.5, y_max, 1.0), minor=True)
ax.grid(True, which='major',alpha = 0.6)
ax.grid(True, which='minor', linewidth=0.4, linestyle='--',alpha = 0.4)
ax.set_xlim(x_min, x_max)
ax.set_ylim(y_min, y_max)
ax.set_aspect('equal')


```
