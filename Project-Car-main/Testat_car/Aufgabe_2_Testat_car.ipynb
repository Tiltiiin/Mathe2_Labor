```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# --- Kurve berechnen ---
t = np.linspace(0, 8 * np.pi, 5000)
x = np.sqrt(t) * np.cos(t)
y = np.sqrt(t) * np.sin(t)



fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(x, y,color = 'black', linewidth=1)

x_max = 5
x_min = -5
y_max = 5
y_min = -5
ax.set_xticks(np.arange(x_min, x_max + 1, 1))
ax.set_yticks(np.arange(y_min, y_max + 1, 1))
ax.set_xticks(np.arange(x_min + 0.5, x_max, 1.0), minor=True)
ax.set_yticks(np.arange(y_min + 0.5, y_max, 1.0), minor=True)
ax.grid(True, which='major',alpha = 0.6)
ax.grid(True, which='minor', linewidth=0.4, linestyle='--',alpha = 0.4)
ax.set_xlim(x_min, x_max)
ax.set_ylim(y_min, y_max)
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
```
