```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
t = np.linspace(0, 8*np.pi, 1000)

# Parametrische Kurve
x = np.sqrt(t) * np.cos(t)
y = np.sqrt(t) * np.sin(t)

# Plot
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(x, y, 'k-')
ax.set_aspect('equal', adjustable='box')
ax.set_xticks(np.arange(-5.001, 5.001, 1))
ax.set_yticks(np.arange(-5.001, 5.001, 1))
ax.set_xticks(np.arange(-5.001, 5.001, 0.5), minor=True)
ax.set_yticks(np.arange(-5.001, 5.001, 0.5), minor=True)
ax.set_xlim(-5.001, 5.001)
ax.set_ylim(-5.001, 5.001)

ax.grid(True)

plt.show()
```
