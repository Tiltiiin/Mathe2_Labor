```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))

# Punkte
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([0, 0, 1, 0, 0], dtype=float)

# Koeffizientenmatrix (Vandermonde-Matrix)
A = np.vstack([x**4, x**3, x**2, x, np.ones(len(x))]).T

# Lösen des linearen Gleichungssystems
coeffs = np.linalg.solve(A, y)
a, b, c, d, e = coeffs

# Polynom auswerten
x_plot = np.linspace(0.5, 5.5, 400)
y_plot = a*x_plot**4 + b*x_plot**3 + c*x_plot**2 + d*x_plot + e

# Plot
ax.plot(x_plot, y_plot, color='blue')
ax.plot(x, y, 'r*', markersize=10)
ax.set_aspect('equal', adjustable='box')
ax.set_xlim(0, 6.5)
ax.set_ylim(-1, 2.5)
ax.set_xticks(np.arange(0, 6.5, 1))
ax.set_yticks(np.arange(-1, 2.5, 1))
ax.set_xticks(np.arange(-0.01, 6.51, 0.5), minor=True)
ax.set_yticks(np.arange(-1.01, 2.51, 0.5), minor=True)

ax.grid(True, which='major')

plt.show()
```
