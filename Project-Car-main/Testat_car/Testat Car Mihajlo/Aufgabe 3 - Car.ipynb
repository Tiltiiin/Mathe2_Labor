```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([
    [-3, -3],
    [0,  6],
    [ 3, -3],
    [np.nan, np.nan],
    [-2, 0],
    [2, 0]
])

# Rotationswinkel
theta = np.pi / 9

# Rotationsmatrix
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

# Rotation
A_rot = A @ R.T

# Plot
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_aspect('equal', adjustable='box')
ax.set_xticks(np.arange(-4.0001, 4.5001, 1))
ax.set_yticks(np.arange(-5.001, 7.501, 1))
ax.set_xticks(np.arange(-4.0001, 4.5001, 0.5), minor=True)
ax.set_yticks(np.arange(-5.001, 7.501, 0.5), minor=True)
ax.set_xlim(-4.001, 4.501)
ax.set_ylim(-5.001, 7.501)
ax.plot(A[:, 0], A[:, 1], 'k-')
ax.plot(A_rot[:, 0], A_rot[:, 1], 'r-')
ax.grid(True)

plt.show()
```
