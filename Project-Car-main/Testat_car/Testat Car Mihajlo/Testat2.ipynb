```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))

# Punkte
x = np.array([1, 2, 3], dtype=float)
y = np.array([1, 2, 2], dtype=float)

# Koeffizientenmatrix
A = np.array([
    [1**2, 1, 1],
    [2**2, 2, 1],
    [3**2, 3, 1],
], dtype=float)

# Lösen des linearen Gleichungssystems
coeffs = np.linalg.solve(A, y)
a, b, c = coeffs

# Polynom auswerten
x_plot = np.linspace(0, 5, 400)
y_plot = a*x_plot**2 + b*x_plot + c

# Plot
ax.plot(x_plot, y_plot, color='blue')
ax.plot(x, y, 'r*', markersize=10)
ax.set_aspect('equal', adjustable='box')
ax.set_xlim(0, 4.5)
ax.set_ylim(0, 3.5)
ax.set_xticks(np.arange(0, 4.5, 1))
ax.set_yticks(np.arange(0, 3.5, 1))
ax.set_xticks(np.arange(-0.01, 4.51, 0.5), minor=True)
ax.set_yticks(np.arange(-0.01, 3.51, 0.5), minor=True)

ax.grid(True, which='major')
ax.grid(True, which='minor', alpha=0.2)

plt.show()
```
