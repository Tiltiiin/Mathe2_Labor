```python
import numpy as np
import matplotlib.pyplot as plt

```

```python

x_max = 6.5
x_min = 0
y_max = 2.5
y_min = -1
grad_funktion = 5
x_pts = np.array([1, 2, 3, 4, 5])
y_pts = np.array([0, 0, 1, 0, 0])
A = np.vander(x_pts, grad_funktion)
coeffs = np.linalg.solve(A, y_pts)
x = np.linspace(x_min, x_max, 1000)
y = np.polyval(coeffs, x)
fig, ax = plt.subplots(figsize=(10, 10))
ax.plot(x, y, 'b-', linewidth=2)
ax.plot(x_pts, y_pts, 'r*', markersize=14)

ax.set_xticks(np.arange(x_min, x_max + 1, 1))
ax.set_yticks(np.arange(y_min, y_max, 1))
ax.set_xticks(np.arange(x_min + 0.5, x_max, 1.0), minor=True)
ax.set_yticks(np.arange(y_min + 0.5, y_max, 1.0), minor=True)
ax.grid(True, which='major',alpha = 0.6)
ax.grid(True, which='minor', linewidth=0.4, linestyle='--',alpha = 0.4)
ax.set_xlim(x_min, x_max)
ax.set_ylim(y_min, y_max)
ax.set_aspect('equal')   
plt.tight_layout()
plt.show()
```
