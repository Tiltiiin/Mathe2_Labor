```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Parameter & Anfangsbedingungen

```python
# Modellparameter
B = 1.0
L0 = 0.5

# Anfangswerte
x0 = 0.6       # x(0)
v0 = 0.0       # x'(0)

# Simulationsparameter
h = 0.001      # Schrittweite
T = 10.0       # Simulationszeit

# Herzschlag-Modell
def acceleration(x, v, B, L0):
    return -100 * (v*x**2 - B*v + x - L0)
```

## 2. Euler-Verfahren

```python
N = int(T / h)
t = np.linspace(0, T, N+1)

x = np.zeros(N+1)
v = np.zeros(N+1)

x[0] = x0
v[0] = v0

for n in range(N):
    x[n+1] = x[n] + h * v[n]
    v[n+1] = v[n] + h * acceleration(x[n], v[n], B, L0)
```

## 3. Plots

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(t, x)
ax1.set_xlabel("Zeit t [s]")
ax1.set_ylabel("Herzmuskel-Länge x(t)")
ax1.set_title("Simulation des Herzschlags")
ax1.grid(False)

ax2.plot(x, v)
ax2.set_xlabel("x (Länge des Herzmuskels)")
ax2.set_ylabel("x' (Geschwindigkeit)")
ax2.set_title("Phasenporträt")
ax2.grid(False)

plt.tight_layout()
plt.show()
```
